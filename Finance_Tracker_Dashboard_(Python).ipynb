## **Create Visualizations**

```python
# Monthly Expenses

monthly_expense = df.groupby('Month')['amount'].sum()

monthly_expense.plot(kind='bar')
plt.title('Monthly Expense')
plt.show()
```

```python
# Expense by Category in Pie chart

category_expense = df.groupby('category')['amount'].sum()

plt.figure(figsize=(10,10))

# Calculate percentages for legend
total_amount = category_expense.sum()
percentages = (category_expense / total_amount * 100).round(1)
legend_labels = [f'{cat} ({p:.1f}%)' for cat, p in zip(category_expense.index, percentages)]

category_expense.plot(
    kind='pie',
    autopct='%1.1f%%',
    startangle=90,
    pctdistance=0.9,
    textprops={'fontsize': 8} # Increased font size for clarity
)

plt.legend(legend_labels, loc="best", fontsize=10) # Updated legend with percentages
plt.title("Expense by Category", fontsize=18)
plt.ylabel("")
plt.show()
```

```python
# Expense by Category in Bar chart

category_expense = df.groupby('category')['amount'].sum()

category_expense.plot(kind='bar', color='skyblue')
plt.title('Expense by Category')
plt.show()
```

```python
# Income vs Expense

df.groupby('Month')[['Income','amount']].sum().plot()
plt.title("Income vs Amount")
plt.show()
```

# **Create Interactive Dashboard**

```python
fig = px.bar(df, x='Month', y='amount', color='category',
             title="Monthly Expense Dashboard")

fig.show()
```

```python
fig = px.pie(df, names='category', values='amount',
             title='Expense Distribution')

fig.show()
```

## Key Insights

• The majority of expenses fall under the **Other** category.

• **Office Supplies** is the second highest spending category.

• Categories like **Transportation, Travel, and Utilities** contribute very little to the total expense.

• Most spending is concentrated in only a few categories.

```python
top_category = df.groupby('category')['amount'].sum().idxmax()
top_value = df.groupby('category')['amount'].sum().max()

print("Key Insights")
print("Highest spending category:", top_category)
print("Total expense in this category:", top_value)
```

```python
total_expense = df['amount'].sum()

print("Total Expense:", total_expense)

avg_expense = df['amount'].mean()
print("Average Expense:", avg_expense)
```

## Key Insights

• The highest expense category is **Other**, contributing more than 80% of total spending.

• **Office Supplies** is the second largest expense category.

• Small categories such as Healthcare, Entertainment, and Transportation contribute less than 5%.

• Expense distribution is highly concentrated in a few categories.

```python
# Total Expense
total_expense = df['amount'].sum()

# Average Expense
average_expense = df['amount'].mean()

# Highest Expense Category
category_expense = df.groupby('category')['amount'].sum()
top_category = category_expense.idxmax()
top_value = category_expense.max()

# Lowest Expense Category
lowest_category = category_expense.idxmin()
lowest_value = category_expense.min()

print("----- Key Insights -----\n")

print(f"Total Expense: {total_expense}")

print(f"Average Expense per Transaction: {average_expense:.2f}")

print(f"Highest Spending Category: {top_category} ({top_value})")

print(f"Lowest Spending Category: {lowest_category} ({lowest_value})")
```
